```python
import numpy as np
import matplotlib.pyplot as plt
from multispec_fitting import *
from lmfit import minimize
```

```python
input_file = 'n2q_input.txt'
bands,amps,constraints = read_params(input_file)
params = parameter_gen(bands,amps,constraints)
```

```python
x0, data, labels = load_data('N2Q_MCD_VTVH.txt')
```

```python
out = minimize(objective, params, args=(x0, bands, data), method = 'least_squares')
outbands, outamps = output_lists(out,bands,amps)
```

```python
upper_range = 28000
lower_range = 9000
color = plt.cm.tab10(np.linspace(0,1,10))
for i in range(len(data)):
    allbands = spec_bands(bands,out.params,x0,i)
    fig,ax = plt.subplots()
    ax.plot(x0,data[i],linewidth = 0.9,color = 'darkgrey')
    ax.plot(x0,spec_model(bands,out.params,x0,i),color = 'black',linewidth = 0.9)
    ax.plot([upper_range,max((lower_range,x0.min()))],[0,0],linestyle='--',color='grey',linewidth = 0.9)
    for j in range(len(allbands)):
        if j == len(allbands)-1:
            ax.plot(x0,allbands[j],color = 'maroon', linestyle = '--', linewidth = 0.8)
        else:
            ax.plot(x0,allbands[j],color = color[j], linestyle = '--', linewidth = 0.8)
    ax.set_xlim(upper_range,max((lower_range,x0.min())))
    ax.set_xlabel("Energy ($cm^{-1}$)")
    ax.set_ylabel("MCD Intensity")

    if 'labels' in locals() and i < len(labels):
        ax.set_title(f"{labels[i]}")
    else:
        ax.set_title(f"MCD Fit Spectrum {i}")

    plt.show()
```
